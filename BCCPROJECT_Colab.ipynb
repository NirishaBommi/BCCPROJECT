# Google Colab Runner for BCC Skin Lesion Detection Dashboard

This notebook guides you through setting up and running the BCC Skin Lesion Detection interactive web dashboard directly on Google Colab **without requiring Google Drive**.

### How to use this notebook:
1. Run the cells in order.
2. In Step 1, select and upload the lightweight project zip file (**`BCCPROJECT_code_only.zip`**) directly from your computer.
3. Click the generated proxy URL at the end of the notebook to open the interactive dashboard.

## Step 1: Upload the Project Zip File
Run this cell to select and upload the lightweight project zip file (**`BCCPROJECT_code_only.zip`**) from your local computer.

```python
from google.colab import files
import os

print("Please select and upload your project zip file (e.g. BCCPROJECT_code_only.zip):")
uploaded = files.upload()
zip_name = list(uploaded.keys())[0]
print(f"\nSuccessfully uploaded: {zip_name}")
```

## Step 2: Unzip the Project Files

```python
import zipfile
import os

dest_dir = '/content/BCCPROJECT'
!mkdir -p {dest_dir}

# Check all possible locations where the zip file might have been uploaded
possible_paths = [
    zip_name,
    os.path.join('/content', zip_name),
    os.path.join('/content/BCCPROJECT', zip_name)
]

zip_path = None
for p in possible_paths:
    if os.path.exists(p):
        zip_path = p
        break

if zip_path is None:
    raise FileNotFoundError(f"Could not find the uploaded zip file '{zip_name}' in any location.")

print(f"Extracting {zip_path} to {dest_dir}...")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(dest_dir)

print("Extraction complete!")

# Change directory to the extracted project folder
%cd {dest_dir}
```
